# 1. `len()`

## Purpose

`len()` returns the number of elements in a sequence.

This is very common in algorithms because we often need to know how many elements are in a list, string, or dictionary.

## Example 1: List

A list stores multiple values in order.

````python
nums = [2, 7, 11, 15]

len(nums)
````

Expected output:

```text
4
```

## Example 2: String

A string is also a sequence.

````python
name = "hello"

len(name)
````

Expected output:

```text
5
```

## Example 3: Dictionary

For a dictionary, `len()` returns the number of key-value pairs.

````python
d = {1: "a", 2: "b"}

len(d)
````

Expected output:

```text
2
```

## Common Usage

In algorithms, I often save the length as `n`.

````python
nums = [2, 7, 11, 15]
n = len(nums)

n
````

# 2. `len()` + `range()`

This is one of the most common algorithm patterns.

`range(len(nums))` creates the valid index positions of the list.

````python
nums = [2, 7, 11, 15]

for i in range(len(nums)):
    print(i)
````

Expected output:

```text
0
1
2
3
```

Here:

- `i` is the index
- `nums[i]` is the value at that index

````python
for i in range(len(nums)):
    print(i, nums[i])
````

Expected output:

```text
0 2
1 7
2 11
3 15
```

# 3. `enumerate()`

`enumerate()` is very common in LeetCode.

Instead of writing:

```python
for i in range(len(nums)):
    print(i, nums[i])
```

Python provides a cleaner way.

````python
for i, num in enumerate(nums):
    print(i, num)
````

Expected output:

```text
0 2
1 7
2 11
3 15
```

`enumerate()` gives:

```text
(index, value)
```

This is cleaner and often preferred.

# 4. Nested Loops

## Purpose

Nested loops are useful when I need to compare every pair.

This often appears in brute-force solutions.

````python
nums = [2, 7, 11, 15]

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        print(nums[i], nums[j])
````

Expected output:

```text
2 7
2 11
2 15
7 11
7 15
11 15
```

This avoids comparing a number with itself.

It also avoids duplicate pairs such as `(2, 7)` and `(7, 2)`.

# 5. Common Loop Patterns

## Pattern 1: Loop over values

````python
for num in nums:
    print(num)
````

Use this when I only need the value and do not need the index.

## Pattern 2: Loop over indices

````python
for i in range(len(nums)):
    print(i, nums[i])
````

Use this when I need to access values by position, such as `nums[i]`.

## Pattern 3: Loop over index and value

````python
for i, num in enumerate(nums):
    print(i, num)
````

This is often the cleanest pattern for LeetCode.

## Pattern 4: Compare every pair

````python
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        print(i, j, nums[i], nums[j])
````

This is the basic brute-force pattern for problems like Two Sum.

## Pattern 5: Reverse loop

````python
for i in range(len(nums) - 1, -1, -1):
    print(i)
````

Expected output:

```text
3
2
1
0
```

## Pattern 6: Sliding window skeleton

````python
left = 0

for right in range(len(nums)):
    print("left:", left, "right:", right)
````

This pattern is very common in medium problems.

# 6. Complexity Intuition

## One loop

```python
for i in range(len(nums)):
    ...
```

Time complexity:

```text
O(n)
```

## Nested loop

```python
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        ...
```

Time complexity:

```text
O(n²)
```

Reason: for each element, I may compare it with many other elements.

## Binary search

```python
while left <= right:
    ...
```

Time complexity:

```text
O(log n)
```

Reason: each step removes about half of the search space.
